```python
# %% [markdown]
# SECTION 1 — Install and import packages

# %%
!pip -q install mat73

import os
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy import io as sio
from scipy import signal
from google.colab import files, drive

# Mount Google Drive so all figures/results are saved permanently.
drive.mount('/content/drive')

RESULTS_DIR = (
    '/content/drive/MyDrive/'
    'NDTsuperresolutiondataverasonicscodeandpostprocessingcode/'
    'NDTsuperresolutiondata/PAUT/Results'
)
os.makedirs(RESULTS_DIR, exist_ok=True)

EPS = np.finfo(np.float64).eps

print('Packages ready.')
print('All figures will be saved in:')
print(RESULTS_DIR)

# %% [markdown]
# SECTION 2 — Upload your PAUT .mat data file

# %%
uploaded = files.upload()
if len(uploaded) == 0:
    raise RuntimeError('No MAT file was uploaded.')

data_file = next(iter(uploaded.keys()))
print('Using file:', data_file)

# %% [markdown]
# SECTION 3 — Helper functions

# %%
def load_mat_any(path):
    """Load MATLAB <=v7.2 with scipy; fall back to mat73 for MATLAB v7.3."""
    try:
        data = sio.loadmat(
            path,
            squeeze_me=True,
            struct_as_record=False,
            simplify_cells=True,
        )
        data = {k: v for k, v in data.items() if not k.startswith('__')}
        print('Loaded with scipy.io.loadmat.')
        return data
    except (NotImplementedError, ValueError, TypeError, OSError) as exc:
        print('scipy.io.loadmat could not read this file as a classic MAT file.')
        print('Trying mat73 for MATLAB v7.3 ...')
        import mat73
        data = mat73.loadmat(path)
        print('Loaded with mat73.')
        return data


def mget(obj, field):
    """Read a MATLAB-struct field from dict-like or attribute-like objects."""
    if isinstance(obj, dict):
        if field not in obj:
            raise KeyError(f'Field {field!r} not found. Available: {list(obj.keys())}')
        return obj[field]
    if hasattr(obj, field):
        return getattr(obj, field)
    raise KeyError(f'Cannot read field {field!r} from object of type {type(obj)}')


def has_field(obj, field):
    if isinstance(obj, dict):
        return field in obj
    return hasattr(obj, field)


def scalar(x):
    a = np.asarray(x)
    if a.size != 1:
        raise ValueError(f'Expected a scalar but got shape {a.shape}')
    return a.reshape(-1)[0].item()


def first_struct(obj):
    """Return the first struct when a singleton struct is wrapped in a list/object array."""
    if isinstance(obj, (list, tuple)):
        return obj[0]
    if isinstance(obj, np.ndarray) and obj.dtype == object:
        return obj.reshape(-1)[0]
    return obj


def struct_array_length(obj, preferred_field=None):
    """Best-effort length of a MATLAB struct array loaded by scipy or mat73."""
    if isinstance(obj, (list, tuple)):
        return len(obj)

    if isinstance(obj, np.ndarray) and obj.dtype == object:
        return obj.size

    if isinstance(obj, dict):
        fields = [preferred_field] if preferred_field and preferred_field in obj else []
        fields += [k for k in obj.keys() if k not in fields]

        for f in fields:
            v = obj[f]
            if isinstance(v, (list, tuple)) and len(v) > 1:
                return len(v)

            arr = np.asarray(v)
            # A scalar field repeated across structs is the safest length indicator.
            if arr.ndim == 1 and arr.size > 1:
                return arr.size
            if arr.ndim == 2 and 1 in arr.shape and arr.size > 1:
                return arr.size

    return 1


def struct_field_at(obj, field, index, n_structs=None):
    """Read field `field` from element `index` of a MATLAB struct array."""
    if isinstance(obj, (list, tuple)):
        return mget(obj[index], field)

    if isinstance(obj, np.ndarray) and obj.dtype == object:
        item = obj.reshape(-1, order='F')[index]
        return mget(item, field)

    if not isinstance(obj, dict):
        # scipy mat_struct array/object fallback
        arr = np.asarray(obj, dtype=object)
        if arr.size > 1:
            return mget(arr.reshape(-1, order='F')[index], field)
        return mget(obj, field)

    v = obj[field]

    if n_structs is None:
        n_structs = struct_array_length(obj)

    if n_structs == 1:
        return v

    if isinstance(v, (list, tuple)):
        if len(v) == n_structs:
            return v[index]

    arr = np.asarray(v)

    # Scalar field for each struct.
    if arr.size == n_structs and (arr.ndim == 1 or 1 in arr.shape):
        return arr.reshape(-1, order='F')[index]

    # Vector/matrix field for each struct. Handle common orientations.
    if arr.ndim >= 2:
        if arr.shape[0] == n_structs:
            return arr[index, ...]
        if arr.shape[-1] == n_structs:
            return arr[..., index]
        if arr.ndim >= 2 and arr.shape[1] == n_structs:
            return arr[:, index, ...]

    # Object arrays from HDF5 readers.
    if arr.dtype == object and arr.size >= n_structs:
        return arr.reshape(-1, order='F')[index]

    raise ValueError(
        f'Could not index field {field!r} as a {n_structs}-element struct array. '
        f'Loaded field shape = {arr.shape}'
    )


def matlab_cell_first(x):
    """Equivalent of MATLAB RcvData{1}."""
    if isinstance(x, (list, tuple)):
        return np.asarray(x[0])

    arr = np.asarray(x)
    if arr.dtype == object:
        return np.asarray(arr.reshape(-1, order='F')[0])
    return arr


def unique_stable(values, decimals=10):
    out = []
    for value in np.round(np.asarray(values, dtype=float).ravel(), decimals):
        if not any(np.isclose(value, old, atol=10**(-decimals)) for old in out):
            out.append(float(value))
    return np.asarray(out, dtype=float)


def sample_complex_trace(rf, sample_index_matlab):
    """
    Complex linear interpolation using MATLAB-style 1-based sample indices.
    This matches the MATLAB sampleComplexTrace() function.
    """
    rf = np.asarray(rf)
    s = np.asarray(sample_index_matlab, dtype=np.float64)

    i0 = np.floor(s).astype(np.int64)       # MATLAB 1-based lower index
    alpha = s - i0

    out = np.zeros(s.shape, dtype=np.complex128)
    valid = np.isfinite(s) & (i0 >= 1) & (i0 < rf.size)

    py0 = i0[valid] - 1                     # Python 0-based lower index
    a = alpha[valid]
    out[valid] = rf[py0] * (1.0 - a) + rf[py0 + 1] * a
    return out


def sample_complex_columns(rf_matrix, sample_index_matlab):
    """
    Interpolate one sample from every RF column.
    rf_matrix: [nSamples, nChannels]
    sample_index_matlab: [nChannels], MATLAB-style 1-based indices
    """
    rf_matrix = np.asarray(rf_matrix)
    s = np.asarray(sample_index_matlab, dtype=np.float64).ravel()
    n_samples, n_channels = rf_matrix.shape

    if s.size != n_channels:
        raise ValueError('sample_index_matlab must have one value per RF channel.')

    i0 = np.floor(s).astype(np.int64)
    alpha = s - i0
    out = np.zeros(n_channels, dtype=np.complex128)
    valid = np.isfinite(s) & (i0 >= 1) & (i0 < n_samples)

    cols = np.arange(n_channels)
    py0 = i0[valid] - 1
    c = cols[valid]
    a = alpha[valid]

    out[valid] = (
        rf_matrix[py0, c] * (1.0 - a)
        + rf_matrix[py0 + 1, c] * a
    )
    return out


def show_and_save_linear_image(
    image,
    x_mm,
    z_mm,
    title,
    filename,
    cmap='jet',
    vmin=0.0,
    vmax=1.0,
):
    """Display and save one LINEAR-SCALE reconstruction image."""
    fig = plt.figure(figsize=(9, 7))
    plt.imshow(
        image,
        extent=[x_mm[0], x_mm[-1], z_mm[-1], z_mm[0]],
        aspect='equal',
        cmap=cmap,
        vmin=vmin,
        vmax=vmax,
    )
    plt.xlabel('Lateral (mm)')
    plt.ylabel('Depth (mm)')
    plt.title(title)
    plt.colorbar(label='Normalized linear amplitude')
    plt.tight_layout()

    save_path = os.path.join(RESULTS_DIR, filename)
    fig.savefig(save_path, dpi=300, bbox_inches='tight')
    print('Saved:', save_path)

    plt.show()
    plt.close(fig)

print('Helper functions ready.')

# %% [markdown]
# SECTION 4 — User settings (same settings as the MATLAB code)

# %%
frame_num = 1                 # MATLAB-style frame number: first frame = 1

auto_calibrate_using_media = True
manual_sample_offset = 0
manual_reverse_rx = False

offset_candidates = np.arange(-30, 31, 1)
use_tx_coherence_weight = True

use_sector_mask = False
sector_half_width_deg = 9.0

save_reconstruction = True
output_file = os.path.join(
    RESULTS_DIR,
    'CustomDAS_Sliding8TX_32RX.mat'
)

print('Settings ready.')

# %% [markdown]
# SECTION 5 — Load the MAT file and read Verasonics parameters

# %%
rawdata = load_mat_any(data_file)
print('Variables in MAT file:')
print(rawdata.keys())

required_fields = ['Trans', 'Resource', 'Receive', 'TX', 'PData', 'RcvData']
missing = [name for name in required_fields if name not in rawdata]
if missing:
    raise KeyError(f'Missing required variables: {missing}')

Trans = first_struct(rawdata['Trans'])
Resource = first_struct(rawdata['Resource'])
Receive = rawdata['Receive']
TX = rawdata['TX']
PData = first_struct(rawdata['PData'])

num_elements = int(round(float(scalar(mget(Trans, 'numelements')))))
num_rx = num_elements

# TX.waveform is scalar for every TX structure and therefore is a good
# field for determining the TX struct-array length.
num_events = struct_array_length(TX, preferred_field='waveform')

parameters = first_struct(mget(Resource, 'Parameters'))
c = float(scalar(mget(parameters, 'speedOfSound')))
f0 = float(scalar(mget(Trans, 'frequency'))) * 1e6

receive_count = struct_array_length(Receive, preferred_field='acqNum')
first_receive_index = (frame_num - 1) * num_events

if first_receive_index >= receive_count:
    raise IndexError('frame_num is larger than the number of Receive structures available.')

try:
    fs = float(scalar(struct_field_at(
        Receive, 'decimSampleRate', first_receive_index, receive_count
    ))) * 1e6
except Exception as exc:
    raise RuntimeError(
        'Receive.decimSampleRate is missing. Upload the acquired/simulated data MAT '
        'file, not only the setup MAT file.'
    ) from exc

lambda_m = c / f0
lambda_mm = lambda_m * 1e3

# TW.peak in center-frequency periods.
if 'TW' in rawdata and rawdata['TW'] is not None:
    TW = rawdata['TW']

    if isinstance(TW, (list, tuple)):
        n_tw = len(TW)
    elif isinstance(TW, np.ndarray) and TW.dtype == object:
        n_tw = TW.size
    elif isinstance(TW, dict) and 'peak' in TW:
        peak_loaded = TW['peak']
        if isinstance(peak_loaded, (list, tuple)):
            n_tw = max(1, len(peak_loaded))
        else:
            peak_arr = np.asarray(peak_loaded)
            n_tw = peak_arr.size if peak_arr.size > 1 else 1
    else:
        n_tw = 1

    tw_peak_periods = np.zeros(n_tw, dtype=float)
    for iw in range(n_tw):
        try:
            tw_peak_periods[iw] = float(scalar(struct_field_at(TW, 'peak', iw, n_tw)))
        except Exception:
            tw_peak_periods[iw] = 0.0
else:
    tw_peak_periods = np.array([0.0])

# P parameters, if saved.
P = first_struct(rawdata['P']) if 'P' in rawdata else None

if P is not None and has_field(P, 'numSubApertures'):
    num_sub_apertures = int(round(float(scalar(mget(P, 'numSubApertures')))))
else:
    apod0 = np.asarray(struct_field_at(TX, 'Apod', 0, num_events), dtype=float).ravel()
    tx_aperture = int(np.count_nonzero(np.abs(apod0) > 0))
    num_sub_apertures = num_elements - tx_aperture + 1

if P is not None and has_field(P, 'numAngles'):
    num_angles = int(round(float(scalar(mget(P, 'numAngles')))))
else:
    num_angles_float = num_events / num_sub_apertures
    if not np.isclose(num_angles_float, round(num_angles_float)):
        raise RuntimeError('Could not determine the number of steering angles.')
    num_angles = int(round(num_angles_float))

# Event steering angles.
if 'TXAngleDeg' in rawdata:
    event_angle_deg = np.asarray(rawdata['TXAngleDeg'], dtype=float).ravel()
else:
    event_angle_deg = np.zeros(num_events, dtype=float)
    for ev in range(num_events):
        steer = np.asarray(struct_field_at(TX, 'Steer', ev, num_events), dtype=float).ravel()
        event_angle_deg[ev] = np.rad2deg(steer[0])

if event_angle_deg.size != num_events:
    event_angle_deg = event_angle_deg.reshape(-1, order='F')[:num_events]

angle_list = unique_stable(event_angle_deg, decimals=10)

print('\n============================================================')
print('CUSTOM SLIDING-8TX PAUT DAS')
print('============================================================')
print(f'Physical elements          = {num_elements}')
print(f'Receive elements/event     = {num_rx}')
print(f'Transmit events/frame      = {num_events}')
print(f'Sliding aperture positions = {num_sub_apertures}')
print('Steering angles            =', '  '.join(f'{a:.1f}' for a in angle_list), 'deg')
print(f'Speed of sound             = {c:.2f} m/s')
print(f'Frequency                  = {f0/1e6:.4f} MHz')
print(f'Sampling frequency         = {fs/1e6:.4f} MHz')
print(f'Wavelength                 = {lambda_mm:.4f} mm')
print('============================================================')

# Element coordinates. Your Verasonics setup uses Trans.units = 'mm'.
element_pos = np.asarray(mget(Trans, 'ElementPos'), dtype=float)
if element_pos.ndim != 2:
    element_pos = np.squeeze(element_pos)
if element_pos.shape[0] != num_elements and element_pos.shape[1] == num_elements:
    element_pos = element_pos.T

element_x_m = element_pos[:num_elements, 0] * 1e-3
print(f'Array X extent = {element_x_m[0]*1e3:.3f} to {element_x_m[-1]*1e3:.3f} mm')

# %% [markdown]
# SECTION 6 — Load one raw frame and find the 32 physical RX columns

# %%
Rall = matlab_cell_first(rawdata['RcvData'])
Rall = np.asarray(Rall)

if Rall.ndim == 2:
    if frame_num != 1:
        raise IndexError('RcvData contains only one frame.')
    R = Rall.astype(np.float64, copy=False)
elif Rall.ndim == 3:
    frame_index = frame_num - 1
    if frame_index >= Rall.shape[2]:
        raise IndexError('Requested frame_num exceeds available RcvData frames.')
    R = np.asarray(Rall[:, :, frame_index], dtype=np.float64)
else:
    raise ValueError(f'Unexpected RcvData dimensions: {Rall.shape}')

print(f'Raw RcvData frame size = {R.shape[0]} rows x {R.shape[1]} columns')

if R.shape[1] == num_rx:
    active_cols = np.arange(num_rx, dtype=int)
else:
    column_energy = np.sum(np.abs(R) ** 2, axis=0)
    order = np.argsort(column_energy)[::-1]
    active_cols = np.sort(order[:num_rx])

    print(f'RcvData has {R.shape[1]} columns although the probe has {num_rx} elements.')
    print('Selected Python RX columns (0-based):', active_cols)
    print('Same columns in MATLAB numbering (1-based):', active_cols + 1)

    plt.figure(figsize=(9, 4))
    plt.plot(np.arange(1, column_energy.size + 1),
             column_energy / (np.max(column_energy) + EPS), 'o-')
    plt.xlabel('RcvData column')
    plt.ylabel('Normalized energy')
    plt.title('Receive-buffer column energy')
    plt.grid(True)
    plt.tight_layout()

    save_path = os.path.join(
        RESULTS_DIR,
        '01_ReceiveBufferColumnEnergy.png'
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    print('Saved:', save_path)

    plt.show()
    plt.close()

# %% [markdown]
# SECTION 7 — Decode the RF buffer into TIME × TX-event × RX

# %%
first_receive = (frame_num - 1) * num_events
last_receive = first_receive + num_events - 1

if last_receive >= receive_count:
    raise IndexError(
        f'Receive does not contain {num_events} events for frame {frame_num}. '
        'Check numEvents/frame and frame_num.'
    )

samples_per_event = np.zeros(num_events, dtype=int)
start_samples = np.full(num_events, -1, dtype=int)
end_samples = np.full(num_events, -1, dtype=int)
acq_nums = np.arange(1, num_events + 1, dtype=int)

for ev in range(num_events):
    ridx = first_receive + ev
    try:
        start_samples[ev] = int(round(float(scalar(
            struct_field_at(Receive, 'startSample', ridx, receive_count)
        ))))
        end_samples[ev] = int(round(float(scalar(
            struct_field_at(Receive, 'endSample', ridx, receive_count)
        ))))
        samples_per_event[ev] = end_samples[ev] - start_samples[ev] + 1
    except Exception:
        samples_per_event[ev] = 0

    try:
        acq_nums[ev] = int(round(float(scalar(
            struct_field_at(Receive, 'acqNum', ridx, receive_count)
        ))))
    except Exception:
        acq_nums[ev] = ev + 1

if np.all(samples_per_event > 0):
    values, counts = np.unique(samples_per_event, return_counts=True)
    n_samples = int(values[np.argmax(counts)])
else:
    n_samples = int(np.floor(R.shape[0] / num_events))

print('Samples per acquisition =', n_samples)

RFcube = np.zeros((n_samples, num_events, num_rx), dtype=np.float64)

direct_addressing_valid = np.all(start_samples >= 1) and np.all(end_samples <= R.shape[0])
direct_addressing_valid = direct_addressing_valid and np.all(
    (end_samples - start_samples + 1) == n_samples
)
direct_addressing_valid = direct_addressing_valid and (
    np.unique(start_samples).size == num_events
)

if direct_addressing_valid:
    print('RF extraction = Receive.startSample/endSample addresses')

    for ev in range(num_events):
        # MATLAB indices are inclusive and 1-based.
        first_row_py = start_samples[ev] - 1
        last_row_py_exclusive = end_samples[ev]

        acq = acq_nums[ev]
        if acq < 1 or acq > num_events:
            acq = ev + 1
        acq_py = acq - 1

        RFcube[:, acq_py, :] = R[
            first_row_py:last_row_py_exclusive,
            active_cols,
        ]
else:
    slot_length = int(np.floor(R.shape[0] / num_events))
    if slot_length < n_samples:
        raise RuntimeError(
            'Could not decode RcvData. Buffer slot is shorter than the acquisition length.'
        )

    print('RF extraction = fixed acquisition slots')
    print('Rows per slot =', slot_length)

    for ev in range(num_events):
        first_row_py = ev * slot_length
        last_row_py_exclusive = first_row_py + n_samples
        RFcube[:, ev, :] = R[
            first_row_py:last_row_py_exclusive,
            active_cols,
        ]

print('RFcube shape =', RFcube.shape, '(TIME, TX-event, RX)')

# %% [markdown]
# SECTION 8 — Convert RF to analytic data and create the exact PData grid

# %%
print('Creating analytic RF data ...')
RFanalytic = signal.hilbert(RFcube, axis=0)
del RFcube
print('Analytic RF created:', RFanalytic.shape)

pdata_origin = np.asarray(mget(PData, 'Origin'), dtype=float).ravel()
pdata_size = np.asarray(mget(PData, 'Size'), dtype=float).ravel().astype(int)
pdata_delta = np.asarray(mget(PData, 'PDelta'), dtype=float).ravel()

x_wl = pdata_origin[0] + np.arange(pdata_size[1]) * pdata_delta[0]
z_wl = pdata_origin[2] + np.arange(pdata_size[0]) * pdata_delta[2]

x_range = x_wl * lambda_m
z_range = z_wl * lambda_m
X, Z = np.meshgrid(x_range, z_range)
n_z, n_x = X.shape

print(f'Reconstruction grid = {n_z} x {n_x} pixels')
print(f'X = {x_range[0]*1e3:.2f} to {x_range[-1]*1e3:.2f} mm')
print(f'Z = {z_range[0]*1e3:.2f} to {z_range[-1]*1e3:.2f} mm')

# %% [markdown]
# SECTION 9 — Decode every sliding 8-element transmit aperture

# %%
TXfirst = np.zeros(num_events, dtype=int)      # 0-based internally
TXlast = np.zeros(num_events, dtype=int)       # 0-based internally
TXcenterX_m = np.zeros(num_events, dtype=float)

print('First/last active TX element for each focal-law event:')

for ev in range(num_events):
    apod = np.asarray(struct_field_at(TX, 'Apod', ev, num_events), dtype=float).ravel()
    active_tx = np.flatnonzero(np.abs(apod) > 0)

    if active_tx.size == 0:
        raise RuntimeError(f'TX event {ev+1} has no active elements.')

    TXfirst[ev] = active_tx[0]
    TXlast[ev] = active_tx[-1]

    w = np.abs(apod[active_tx])
    TXcenterX_m[ev] = np.sum(w * element_x_m[active_tx]) / (np.sum(w) + EPS)

    print(
        f'Event {ev+1:2d}: angle={event_angle_deg[ev]:6.1f} deg   '
        f'TX={TXfirst[ev]+1:2d}:{TXlast[ev]+1:2d}   '
        f'center={TXcenterX_m[ev]*1e3:7.3f} mm'
    )

# %% [markdown]
# SECTION 10 — Automatic RX-order and sample-offset calibration (simulation)

# %%
best_sample_offset = int(manual_sample_offset)
best_rx_order = np.arange(num_rx, dtype=int)
if manual_reverse_rx:
    best_rx_order = best_rx_order[::-1]

Media = first_struct(rawdata['Media']) if 'Media' in rawdata else None
can_calibrate = (
    auto_calibrate_using_media
    and Media is not None
    and has_field(Media, 'MP')
)

if can_calibrate:
    MP = np.asarray(mget(Media, 'MP'), dtype=float)
    MP = np.squeeze(MP)
    if MP.ndim != 2:
        can_calibrate = False
    elif MP.shape[1] < 3 and MP.shape[0] >= 3:
        MP = MP.T
    if can_calibrate and MP.size == 0:
        can_calibrate = False

if can_calibrate:
    print('\n============================================================')
    print('AUTO-CALIBRATING RX ORDER AND SAMPLE OFFSET')
    print('============================================================')

    true_x_m = MP[:, 0] * lambda_m
    true_z_m = MP[:, 2] * lambda_m
    num_targets = true_x_m.size

    rx_orders = [
        np.arange(num_rx, dtype=int),
        np.arange(num_rx - 1, -1, -1, dtype=int),
    ]
    rx_order_names = ['NORMAL', 'REVERSED']

    calibration_scores = np.zeros((2, offset_candidates.size), dtype=float)

    # Precompute physical RX propagation times for every target.
    rx_time_targets = np.zeros((num_targets, num_rx), dtype=float)
    for ip in range(num_targets):
        rx_time_targets[ip, :] = np.sqrt(
            (true_x_m[ip] - element_x_m) ** 2 + true_z_m[ip] ** 2
        ) / c

    # Precompute the same TX focal-law time and coherence used in the MATLAB code.
    tx_time_targets = np.zeros((num_targets, num_events), dtype=float)
    tx_coherence_targets = np.ones((num_targets, num_events), dtype=float)

    for ev in range(num_events):
        apod = np.asarray(struct_field_at(TX, 'Apod', ev, num_events), dtype=float).ravel()
        delay = np.asarray(struct_field_at(TX, 'Delay', ev, num_events), dtype=float).ravel()
        active_tx = np.flatnonzero(np.abs(apod) > 0)

        tx_weights = np.abs(apod[active_tx])
        tx_weights = tx_weights / (np.sum(tx_weights) + EPS)

        waveform_index_matlab = int(round(float(scalar(
            struct_field_at(TX, 'waveform', ev, num_events)
        ))))
        waveform_index_py = waveform_index_matlab - 1
        if 0 <= waveform_index_py < tw_peak_periods.size:
            event_tw_peak = tw_peak_periods[waveform_index_py]
        else:
            event_tw_peak = 0.0

        tx_delay_sec = (delay[active_tx] + event_tw_peak) / f0

        for ip in range(num_targets):
            tx_dist = np.sqrt(
                (true_x_m[ip] - element_x_m[active_tx]) ** 2
                + true_z_m[ip] ** 2
            )
            tx_path_times = tx_delay_sec + tx_dist / c
            tx_time = np.sum(tx_weights * tx_path_times)
            tx_time_targets[ip, ev] = tx_time

            if use_tx_coherence_weight:
                tx_coherence_targets[ip, ev] = np.abs(np.sum(
                    tx_weights
                    * np.exp(-1j * 2.0 * np.pi * f0 * (tx_path_times - tx_time))
                ))

    for io, rx_order in enumerate(rx_orders):
        for ioff, off in enumerate(offset_candidates):
            target_values = np.zeros(num_targets, dtype=np.complex128)

            for ip in range(num_targets):
                pixel_value = 0.0 + 0.0j

                for ev in range(num_events):
                    sample_index = (
                        tx_time_targets[ip, ev]
                        + rx_time_targets[ip, :]
                    ) * fs + 1.0 + off

                    rf_matrix = RFanalytic[:, ev, :][:, rx_order]
                    event_samples = sample_complex_columns(rf_matrix, sample_index)
                    event_value = np.sum(event_samples)

                    pixel_value += tx_coherence_targets[ip, ev] * event_value

                target_values[ip] = pixel_value

            calibration_scores[io, ioff] = np.mean(np.abs(target_values))

    best_flat = np.argmax(calibration_scores)
    best_order_index, best_offset_index = np.unravel_index(
        best_flat, calibration_scores.shape
    )

    best_rx_order = rx_orders[best_order_index]
    best_sample_offset = int(offset_candidates[best_offset_index])

    print('Best RX order      =', rx_order_names[best_order_index])
    print(f'Best sample offset = {best_sample_offset:+d} samples')
    print(
        'Equivalent two-way depth shift = '
        f'{best_sample_offset*c/(2*fs)*1e3:.3f} mm'
    )
    print('============================================================')

    norm = np.max(calibration_scores) + EPS
    plt.figure(figsize=(8, 5))
    plt.plot(offset_candidates, calibration_scores[0, :] / norm, 'o-', label='RX normal')
    plt.plot(offset_candidates, calibration_scores[1, :] / norm, 's-', label='RX reversed')
    plt.xlabel('Sample offset')
    plt.ylabel('Normalized target response')
    plt.title('Automatic timing / RX-order calibration')
    plt.grid(True)
    plt.legend()
    plt.tight_layout()

    save_path = os.path.join(
        RESULTS_DIR,
        '02_AutomaticTiming_RXOrderCalibration.png'
    )
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    print('Saved:', save_path)

    plt.show()
    plt.close()
else:
    print('Automatic calibration not used.')
    print('RX order      =', 'REVERSED' if manual_reverse_rx else 'NORMAL')
    print(f'Sample offset = {best_sample_offset:+d} samples')

# %% [markdown]
# SECTION 11 — Precompute receive propagation-time maps

# %%
print('Precomputing RX propagation-time maps ...')
rx_time_maps = np.empty((num_rx, n_z, n_x), dtype=np.float64)
for irx in range(num_rx):
    rx_time_maps[irx] = np.sqrt(
        (X - element_x_m[irx]) ** 2 + Z ** 2
    ) / c
print('RX maps ready:', rx_time_maps.shape)

# %% [markdown]
# SECTION 12 — Final custom DAS for sliding 8TX / all 32RX

# %%
print('\n============================================================')
print('STARTING SLIDING-8TX CUSTOM DAS')
print('============================================================')

image_DAS = np.zeros((n_z, n_x), dtype=np.complex128)
image_by_angle = np.zeros((n_z, n_x, angle_list.size), dtype=np.complex128)
events_per_angle = np.zeros(angle_list.size, dtype=int)

start_time = time.time()

for ev in range(num_events):
    print(
        f'TX event {ev+1} / {num_events}   '
        f'angle {event_angle_deg[ev]:.1f} deg   '
        f'aperture {TXfirst[ev]+1}:{TXlast[ev]+1}'
    )

    apod = np.asarray(struct_field_at(TX, 'Apod', ev, num_events), dtype=float).ravel()
    delay = np.asarray(struct_field_at(TX, 'Delay', ev, num_events), dtype=float).ravel()
    active_tx = np.flatnonzero(np.abs(apod) > 0)

    tx_weights = np.abs(apod[active_tx])
    tx_weights = tx_weights / (np.sum(tx_weights) + EPS)

    waveform_index_matlab = int(round(float(scalar(
        struct_field_at(TX, 'waveform', ev, num_events)
    ))))
    waveform_index_py = waveform_index_matlab - 1
    if 0 <= waveform_index_py < tw_peak_periods.size:
        event_tw_peak = tw_peak_periods[waveform_index_py]
    else:
        event_tw_peak = 0.0

    tx_delay_sec = (delay[active_tx] + event_tw_peak) / f0

    # Same focal-law delay-centroid TX time map as MATLAB.
    tx_time = np.zeros((n_z, n_x), dtype=np.float64)
    tx_phasor = np.zeros((n_z, n_x), dtype=np.complex128)

    for it, el in enumerate(active_tx):
        dtx = np.sqrt((X - element_x_m[el]) ** 2 + Z ** 2)
        tau_i = tx_delay_sec[it] + dtx / c
        tx_time += tx_weights[it] * tau_i

    if use_tx_coherence_weight:
        for it, el in enumerate(active_tx):
            dtx = np.sqrt((X - element_x_m[el]) ** 2 + Z ** 2)
            tau_i = tx_delay_sec[it] + dtx / c
            tx_phasor += (
                tx_weights[it]
                * np.exp(-1j * 2.0 * np.pi * f0 * (tau_i - tx_time))
            )
        tx_weight_map = np.abs(tx_phasor)
    else:
        tx_weight_map = np.ones((n_z, n_x), dtype=np.float64)

    # Optional sector mask, exactly matching the MATLAB logic.
    if use_sector_mask:
        pixel_angle_deg = np.rad2deg(np.arctan2(X - TXcenterX_m[ev], Z))
        angle_difference = (
            (pixel_angle_deg - event_angle_deg[ev] + 180.0) % 360.0
        ) - 180.0
        event_mask = np.abs(angle_difference) <= sector_half_width_deg
    else:
        event_mask = np.ones((n_z, n_x), dtype=bool)

    # Receive DAS for this transmit event.
    event_image = np.zeros((n_z, n_x), dtype=np.complex128)

    for irx in range(num_rx):
        rx_raw = best_rx_order[irx]
        tof = tx_time + rx_time_maps[irx]
        sample_index = tof * fs + 1.0 + best_sample_offset

        rf = RFanalytic[:, ev, rx_raw]
        contribution = sample_complex_trace(rf, sample_index)
        event_image += contribution

    event_image /= num_rx
    event_image *= tx_weight_map
    event_image[~event_mask] = 0.0

    image_DAS += event_image

    ia = int(np.argmin(np.abs(angle_list - event_angle_deg[ev])))
    image_by_angle[:, :, ia] += event_image
    events_per_angle[ia] += 1

elapsed_time = time.time() - start_time
print(f'Beamforming completed in {elapsed_time:.2f} s')

# Same linear normalization as MATLAB.
image_DAS /= num_events
DAS_env = np.abs(image_DAS)
DAS_env /= np.max(DAS_env) + EPS

# %% [markdown]
# SECTION 13 — Combined PAUT DAS image — LINEAR SCALE ONLY

# %%
x_mm = x_range * 1e3
z_mm = z_range * 1e3

show_and_save_linear_image(
    DAS_env,
    x_mm,
    z_mm,
    'Custom DAS - Sliding 8TX / All 32RX - Linear Scale',
    '03_Combined_Sliding8TX_All32RX_Linear.png',
    cmap='jet',
    vmin=0.0,
    vmax=1.0,
)

# %% [markdown]
# SECTION 14 — Individual steering-angle reconstructions — LINEAR SCALE ONLY

# %%
for ia, angle in enumerate(angle_list):

    temp_image = image_by_angle[:, :, ia] / max(events_per_angle[ia], 1)

    temp_env = np.abs(temp_image)
    temp_env /= np.max(temp_env) + EPS

    angle_tag = f'{angle:g}'.replace('.', 'p')

    show_and_save_linear_image(
        temp_env,
        x_mm,
        z_mm,
        f'Sliding 8TX Custom DAS - {angle:.1f} deg - Linear Scale',
        f'04_Angle_{angle_tag}deg_Linear.png',
        cmap='jet',
        vmin=0.0,
        vmax=1.0,
    )

# %% [markdown]
# SECTION 15 — FINAL IMAGE: linear-scale DAS overlaid with true scatterers

# %%
if Media is not None and has_field(Media, 'MP'):

    MP = np.asarray(mget(Media, 'MP'), dtype=float)
    MP = np.squeeze(MP)

    if MP.ndim == 2:

        if MP.shape[1] < 3 and MP.shape[0] >= 3:
            MP = MP.T

        true_x_mm = MP[:, 0] * lambda_mm
        true_z_mm = MP[:, 2] * lambda_mm

        fig = plt.figure(figsize=(9, 7))

        plt.imshow(
            DAS_env,
            extent=[x_mm[0], x_mm[-1], z_mm[-1], z_mm[0]],
            aspect='equal',
            cmap='jet',
            vmin=0.0,
            vmax=1.0,
        )

        plt.plot(
            true_x_mm,
            true_z_mm,
            'ro',
            markersize=8,
            markerfacecolor='none',
            markeredgewidth=1.5,
            label='True scatterers',
        )

        plt.xlabel('Lateral (mm)')
        plt.ylabel('Depth (mm)')
        plt.title(
            'FINAL: Sliding 8TX Custom DAS + True Scatterers - Linear Scale'
        )

        plt.colorbar(label='Normalized linear amplitude')
        plt.legend(loc='best')
        plt.tight_layout()

        final_overlay_path = os.path.join(
            RESULTS_DIR,
            '05_FINAL_Combined_Linear_WithTrueScatterers.png'
        )

        fig.savefig(
            final_overlay_path,
            dpi=300,
            bbox_inches='tight'
        )

        print('Saved FINAL scatterer-overlay image:')
        print(final_overlay_path)

        plt.show()
        plt.close(fig)

    else:
        print('Media.MP exists but its shape could not be interpreted.')
else:
    print(
        'Media.MP is not available, so the final true-scatterer overlay '
        'cannot be generated for this dataset.'
    )

# %% [markdown]
# SECTION 16 — Save the reconstruction as a MATLAB-compatible .mat file

# %%
if save_reconstruction:
    # Save RX order in MATLAB-style 1-based indexing for convenience.
    sio.savemat(
        output_file,
        {
            'image_DAS': image_DAS,
            'imageByAngle': image_by_angle,
            'DAS_env': DAS_env,
            'x_range': x_range,
            'z_range': z_range,
            'X': X,
            'Z': Z,
            'angleList': angle_list,
            'eventAngleDeg': event_angle_deg,
            'TXfirst': TXfirst + 1,
            'TXlast': TXlast + 1,
            'TXcenterX_m': TXcenterX_m,
            'bestRXorder': best_rx_order + 1,
            'bestSampleOffset': best_sample_offset,
            'fs': fs,
            'f0': f0,
            'c': c,
            'frameNum': frame_num,
        },
        do_compression=True,
    )

    print('Saved reconstruction to Google Drive:')
    print(output_file)

print('\n============================================================')
print('DONE')
print('============================================================')
print('Results folder:')
print(RESULTS_DIR)
print('\nImages are LINEAR SCALE only.')
print('The final scatterer-overlay image is:')
print(
    os.path.join(
        RESULTS_DIR,
        '05_FINAL_Combined_Linear_WithTrueScatterers.png'
    )
)
```
